```python
import numpy as np
import matplotlib.pyplot as plt


X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])


#setting up the polynomial
degree = 3
coefficients = np.polyfit(X, Y, degree)
polynomial = np.poly1d(coefficients)
y_pred = polynomial(X)



A = np.vstack([X, np.ones(len(X))]).T

m, c = np.linalg.lstsq(A, Y, rcond=None)[0]


plt.figure()
plt.scatter(X,Y)
plt.plot(X, m*X + c)
plt.plot(X, y_pred, color='red', label = "polynomial fit")
plt.title("Polynomial and Least Squares Fit")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()

y_ls = m*X + c

# figuring out the mean squared error
mse_ls = np.mean((Y - y_ls)**2)
mse_poly = np.mean((Y - y_pred)**2)

# figuring out the r squareed
ss_tot = np.sum((Y - np.mean(Y))**2)

r2_ls = 1 - np.sum((Y - y_ls)**2)/ss_tot
r2_poly = 1 - np.sum((Y - y_pred)**2)/ss_tot

print("Least Squares MSE:", mse_ls)
print("Polynomial MSE:", mse_poly)
print("Least Squares R^2:", r2_ls)
print("Polynomial R^2:", r2_poly)

if mse_ls < mse_poly:
    print("Least Squares is better")
else:
    print("Polynomial is better")

```

Iterating and increasing the degree to see when it significantly overfits.

starting with degree 3

```python
X = np.sort(np.random.rand(10) * 10)
Y = np.random.rand(10) * 10

for i in range(8):
    #setting up the polynomial
    degree = i + 3
    print("New iteration")
    print(f"Polynomial degree: {degree}")
    coefficients = np.polyfit(X, Y, degree)
    polynomial = np.poly1d(coefficients)
    y_pred = polynomial(X)

    A = np.vstack([X, np.ones(len(X))]).T

    m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

    plt.figure()
    plt.scatter(X,Y)
    plt.plot(X, m*X + c)
    plt.plot(X, y_pred, color='red', label = "polynomial fit")
    plt.title("Polynomial and Least Squares Fit")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.show()

    y_ls = m*X + c

    mse_poly = np.mean((Y - y_pred)**2)
    mse_ls = np.mean((Y - y_ls)**2)

    ss_tot = np.sum((Y - np.mean(Y))**2)

    r2_poly = 1 - np.sum((Y - y_pred)**2)/ss_tot
    r2_ls = 1 - np.sum((Y - y_ls)**2)/ss_tot

    print("Least Squares MSE:", mse_ls)
    print("Polynomial MSE:", mse_poly)
    print("Least Squares R^2:", r2_ls)
    print("Polynomial R^2:", r2_poly)

    if r2_poly > 0.99:
        print("Result: Likely overfitting")
    elif r2_poly < 0.5:
        print("Result: Underfitting")
    else:
        print("Result: Reasonable fit")
```

Iterating to see when starts to underfit significantly
starting with degree 5

```python
X = np.sort(np.random.rand(10) * 10)
Y = np.random.rand(10) * 10

degree = 5

while degree > 0:
    #setting up the polynomial
    print("New iteration")
    print(f"Polynomial degree: {degree}")
    coefficients = np.polyfit(X, Y, degree)
    polynomial = np.poly1d(coefficients)
    y_pred = polynomial(X)

    A = np.vstack([X, np.ones(len(X))]).T

    m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

    plt.figure()
    plt.scatter(X,Y)
    plt.plot(X, m*X + c)
    plt.plot(X, y_pred, color='red', label = "polynomial fit")
    plt.title("Polynomial and Least Squares Fit")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.show()

    y_ls = m*X + c

    mse_poly = np.mean((Y - y_pred)**2)
    mse_ls = np.mean((Y - y_ls)**2)

    ss_tot = np.sum((Y - np.mean(Y))**2)

    r2_poly = 1 - np.sum((Y - y_pred)**2)/ss_tot
    r2_ls = 1 - np.sum((Y - y_ls)**2)/ss_tot

    print("Least Squares MSE:", mse_ls)
    print("Polynomial MSE:", mse_poly)
    print("Least Squares R^2:", r2_ls)
    print("Polynomial R^2:", r2_poly)

    if r2_poly > 0.99:
        print("Result: Likely overfitting")
    elif r2_poly < 0.5:
        print("Result: Underfitting")
    else:
        print("Result: Reasonable fit")
    degree -= 1
```

```python
# doing the vandermonde fit

X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

degree = 3

V = np.vander(X, N=degree+1, increasing=True)

VT_V = np.dot(V.T, V)
VT_Y = np.dot(V.T, Y)

coeff_vander = np.linalg.solve(VT_V, VT_Y)

y_vander = V @ coeff_vander


# calculating the mean squared error
mse_vander = np.mean((Y - y_vander)**2)

# calculating the r squared
ss_tot = np.sum((Y - np.mean(Y))**2)
r2_vander = 1 - np.sum((Y - y_vander)**2)/ss_tot

plt.figure()
plt.scatter(X,Y)
plt.plot(X, y_vander)
plt.title("Vandermonde polynomial fit")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()


print("Vandermonde MSE:", mse_vander)
print("Vandermonde R^2:", r2_vander)
```
