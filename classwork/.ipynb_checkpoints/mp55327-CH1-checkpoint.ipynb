do 4 diff types of plots
get them on the same plot as subplots

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 1000, 1000)
y = x**2
plt.semilogy(x, y)
plt.show()
```

```python
dataaa = np.random.rand(1000)
plt.hist(dataaa, bins=50)
plt.title("Hist")
plt.show()
```

```python
lab = ["B", "K", "P"]
val = [10, 5, 8]
plt.bar(lab, val)
plt.title("bar chartt")
plt.show()
```

```python
a = np.random.rand(100,100)
plt.imshow(a)
plt.colorbar()
plt.show()
```

```python
plt.figure()
plt.subplot(2, 2, 1)

plt.semilogy(x, y)
plt.subplot(2, 2, 2)

plt.hist(dataaa, bins=50)
plt.subplot(2, 2, 3)

plt.bar(lab, val)
plt.subplot(2, 2, 4)

plt.imshow(a)
plt.colorbar()

plt.show()
```
