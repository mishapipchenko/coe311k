```python
import numpy as np

def diagonally_dominant(a):
    diag = np.abs(np.diag(a))
    row_sums = np.sum(np.abs(a), axis=1) - diag
    return np.all(diag > row_sums)
```

```python
A = np.array([
    [10, -1, 2, 0, 0],
    [-1, 11, -1, 3, 0],
    [2, -1, 10, -1, 0],
    [0, 3, -1, 8, -2],
    [0, 0, 0, -2, 9]], dtype=float)

print(diagonally_dominant(A))
```

```python
A = np.array([
    [1, 2, 3, 0, 0],
    [2, 1, 2, 3, 0],
    [3, 2, 1, 2, 3],
    [0, 3, 2, 1, 2],
    [0, 0, 3, 2, 1]], dtype=float)

print(diagonally_dominant(A))
```
