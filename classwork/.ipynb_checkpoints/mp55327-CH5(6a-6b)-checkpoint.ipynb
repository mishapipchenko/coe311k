```python
import numpy as np
import matplotlib.pyplot as plt
import time
```

```python
#jacobi method
def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
            
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
            
        x = x_new
        
    return x, max_iterations


#gauss-seidel
def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()

    for k in range(max_iterations):
        x_new = x.copy()

        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i))
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n))
            x_new[i] = (b[i] - s1 - s2) / A[i][i]

        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k

        x = x_new

    return x, max_iterations
```

```python
A = np.array([
    [10, -1, 2, 0, 0],
    [-1, 11, -1, 3, 0],
    [2, -1, 10, -1, 0],
    [0, 3, -1, 8, -2],
    [0, 0, 0, -2, 9]], dtype=float)

b = np.array([14, 30, 26, 25, 37], dtype=float)
x0 = np.zeros_like(b, dtype=float)
tol = 1e-6
max_iterations = 100

start_time = time.time()
solution, iterations = jacobi(A, b, x0, tol, max_iterations)
end_time = time.time()


print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
print(f"Time: {end_time - start_time} seconds")

start_time = time.time()
solution, iterations = gauss_seidel(A, b, x0, tol, max_iterations)
end_time = time.time()


print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
print(f"Time: {end_time - start_time} seconds")
```

Both methods converged. Jacobi took 18 iterations and Gauss Seidel took 10. Jacobi ran in 0.0022873878479003906 seconds, while Gauss Seidel ran quicker, 0.0010821819305419922 seconds

```python
A = np.array([
    [1, 2, 3, 0, 0],
    [2, 1, 2, 3, 0],
    [3, 2, 1, 2, 3],
    [0, 3, 2, 1, 2],
    [0, 0, 3, 2, 1]], dtype=float)


b = np.array([14, 22, 33, 26, 22], dtype=float)
x0 = np.zeros_like(b, dtype=float)
tol = 1e-6
max_iterations = 100

start_time = time.time()
solution, iterations = jacobi(A, b, x0, tol, max_iterations)
end_time = time.time()


print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
print(f"Time: {end_time - start_time} seconds")

start_time = time.time()
solution, iterations = gauss_seidel(A, b, x0, tol, max_iterations)
end_time = time.time()


print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
print(f"Time: {end_time - start_time} seconds")
```

Both methods did not converge and the answers became very large. They both hit the max limit of iterations and solutions ended up diverging. Jacobi ran in 0.007279634475708008 seconds, while Gauss Seidel ran quicker in 0.005696535110473633 seconds
