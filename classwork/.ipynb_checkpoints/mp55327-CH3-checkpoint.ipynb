```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

```python
def monte_carlo_pi(samples):
    ins = 0
    outs = 0
    for n in range(samples):
        a = np.random.uniform(-1,1)
        b = np.random.uniform(-1,1)
        if(a**2+b**2 <= 1):
            ins += 1
        else:
            outs += 1
    pi = (4*ins)/samples
    return pi
```

```python
rel_error_list = []
abs_error_list = []

for i in range(100):
    pi_est = monte_carlo_pi(1000)
    abs_error = np.abs(pi_est - np.pi)
    abs_error_list.append(abs_error)
    rel_error = np.abs((np.pi - pi_est)/np.pi)
    rel_error_list.append(rel_error)
    
plt.plot(abs_error_list)
plt.show()

plt.plot(rel_error_list)
plt.show()
```
