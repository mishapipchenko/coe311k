```python
import numpy as np
import time 
import matplotlib.pyplot as plt
```

```python
def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    
    return C
```

```python
naive_t = []
numpy_t = []

matrix_sizes = []

cur_size = 3 

for i in range(8):
    A = np.random.rand(cur_size, cur_size)
    B = np.random.rand(cur_size, cur_size)
    
    matrix_sizes.append(cur_size)
    
    start = time.time()
    naive_matrix_multiplication(A,B)
    naive_t.append(time.time() - start)
    
    start = time.time()
    np.dot(A,B)
    numpy_t.append(time.time() - start)
    cur_size = cur_size * 2
    

plt.yscale('log')

plt.plot(range(1, 9),naive_t, label="Naive")
plt.plot(range(1, 9),numpy_t, label="Numpy")

plt.title('Runtime Naive v Numpy')
plt.xlabel('Matrix Size')
plt.ylabel('time (sec)')
plt.legend()
```

Used the log scale here to properly highlight the differences in runtime
