```python
import numpy as np
import matplotlib.pyplot as plt

# Data points 

X = np.array([1, 2, 3, 4, 5])

Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

# Stack the design matrix with a column of ones (for the intercept)

A = np.vstack([X, np.ones(len(X))]).T

# Solve using least squares

m, c = np.linalg.lstsq(A, Y, rcond=None)[0]


plt.figure()
plt.scatter(X,Y)
plt.plot(X, m*X + c)
plt.title("data points and best fit line")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()

print(f"Slope: {m}, Intercept: {c}")
```

least squares approximation looks good

```python
X = np.sort(np.random.rand(5) * 10)
Y = np.random.rand(5) * 10
A = np.vstack([X, np.ones(len(X))]).T

m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

plt.figure()
plt.scatter(X,Y)
plt.plot(X, m*X + c)
plt.title("data points and best fit line")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()

print(f"Slope: {m}, Intercept: {c}")
```

As only X is sorted, least squares struggles due to Y being randomized. A good linear approximation is not possible with such random y points, and therefore the line is not a good fit for the points.

```python
m = 2
b = 5

x = np.arange(0,1.01, 0.01)

error = np.random.uniform(-1,1, size=len(x))
y = m*x + b + error
y

A = np.vstack([x, np.ones(len(x))]).T
m_f, b_f = np.linalg.lstsq(A, y, rcond=None)[0]
print(m_f, b_f)

y_pred = m_f*x + b_f
res = y - y_pred

print(res)

rmax = np.argmax(res)
rmin = np.argmin(res)

print("Max residual point:")
print("x =", x[rmax], "y =", y[rmax], "residual =", res[rmax])

print("Min residual point:")
print("x =", x[rmin], "y =", y[rmin], "residual =", res[rmin])

plt.scatter(x, y)
plt.plot(x, y_pred, color='red')
plt.scatter(x[rmax], y[rmax])
plt.scatter(x[rmin], y[rmin])
plt.title("data points and best fit line")
plt.xlabel("x")
plt.show()
```

The least squares method well approximates the original function, as can be seen by the asymptote of 5 and the line ending at 7, which is correct according to the original equation. The residuals of 0.91 and -1.05 are also very close to the added noise, so that makes sense.
