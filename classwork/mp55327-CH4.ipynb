leibniz series for pi 

```python
def leibniz_series(iter):
    sum_series = 0
    for i in range(iter):
        num = ((-1)**i)/(2*i+1)
        sum_series +=num
        if i%100 == 0:
            print(np.abs((4*sum_series - np.pi)/np.pi))
        
    pi_ser = 4 * sum_series
    return pi_ser
```

```python
leibniz_series(5000)
```
