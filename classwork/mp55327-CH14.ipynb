```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
f = lambda x: np.sin(x)
a = 0
b = 2 * np.pi
N = 50
x = np.linspace(a, b, N)
```

```python
# Step size
h = (b - a) / (N - 1)
exact_dx = np.cos(x)
```

```python
# Central Difference Approximation
central_dx = (f(x + h) - f(x - h)) / (2 * h)

# Forward Difference Approximation
forward_dx = (f(x + h) - f(x)) / h

# Backward Difference Approximation
backward_dx = (f(x) - f(x - h)) / h
```

```python
plt.plot(x, exact_dx, label='Exact Derivative: cos(x)')
plt.plot(x, forward_dx,label='Forward Difference')
plt.plot(x, backward_dx,label='Backward Difference')
plt.plot(x, central_dx, label='Central Difference')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title("derivative approximations of sin(x))")
```
