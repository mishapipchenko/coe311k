```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# initial state
k = 1.0
x_start = 0
x_end = 10
y_0 = 1.0
v_0 = 0.0
```

```python
h_values = [0.01, 0.02, 0.03, 0.05, 0.1, 0.2, 0.3, 0.5]
```

```python
for h in h_values:
    # setting up the arrays
    x_values = np.arange(x_start, x_end + h, h)
    y_values = np.zeros(len(x_values))
    v_values = np.zeros(len(x_values))
    
    y_values[0] = y_0
    v_values[0] = v_0

    # running euler's
    for i in range(1, len(x_values)):
        y_n = y_values[i-1]
        v_n = v_values[i-1]
        
        y_values[i] = y_n + h * v_n
        v_values[i] = v_n + h * (-k * y_n)
    plt.plot(x_values, y_values, label=f'h = {h}')
    
plt.xlabel('x')
plt.ylabel('position y(x)')
plt.title("euler's method stability check across different h values")
plt.legend()
plt.show()
```

when does it become unstable? Is it becoming unstable?

Yes, it becomes unstable. It is easy to see that the amplitude of the waves is increasing, when in a perfect constant spring mass system would not be behaving like this.

Play with the value of K, what are you seeing?

A higher value of k causes the amplitude to increase. The higher spring constant increases the frequency and that requires a smaller time step to keep errors from compounding.
