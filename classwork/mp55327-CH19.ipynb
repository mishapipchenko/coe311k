```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
h = 0.0005
t = np.arange(0, 0.02, h)
y_euler = np.zeros(len(t))

for i in range(len(t) - 1):
    y_euler[i+1] = y_euler[i] + h * (-1000 * y_euler[i] + 3000 - 2000 * np.exp(-t[i]))

y_analytical = 3 - 0.998 * np.exp(-1000 * t) - 2.002 * np.exp(-t)
```

```python
plt.plot(t, y_euler, label='Euler Forward')
plt.plot(t, y_analytical, label='Analytical')
plt.legend()
plt.show()
```
