Iterate the number of samples for your Monte Carlo Pi approximation from 10 to 1000, in steps of 20, and plot the error (absolute value of your calculated value of Pi minus the actual value of Pi encoded in Python, to 15 digits past the decimal point)

 

Your notebook must contain: your main code, your Monte Carlo Pi function, and your plot routine, plus comments and citations.

```python
import random
import numpy as np
import matplotlib.pyplot as plt
iterations = np.arange(10,1000,20)
error_list = []

def monte_carlo_pi(samples):
    ins = 0
    outs = 0
    for n in range(samples):
        a = np.random.uniform(-1,1)
        b = np.random.uniform(-1,1)
        if(a**2+b**2 <= 1):
            ins += 1
        else:
            outs += 1
    pi = (4*ins)/samples
    return pi



for i in iterations:
    pi_est = monte_carlo_pi(i)
    error = np.abs(pi_est - np.pi)
    error_list.append(error)
    
plt.figure()
plt.plot(iterations, error_list)
plt.xlabel('n_samples')
plt.ylabel('error')
plt.show()
            
        
```
