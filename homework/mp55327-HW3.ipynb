```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
from math import factorial

def sin_taylor_series(n_terms, x_val):
    sum = 0
    for n in range(n_terms):
        sum += ((-1)**n) * x_val**(2*n + 1) / factorial(2*n + 1)
    return sum
        
        
```

```python
x_val = np.arange(0, 10.1, 0.1)
exact = np.sin(x_val)
error1, error2, error3, error4 = [], [], [], []
```

```python
for x in x_val:
    error1.append(abs(np.sin(x) - sin_taylor_series(1, x)))
    error2.append(abs(np.sin(x) - sin_taylor_series(2, x)))
    error3.append(abs(np.sin(x) - sin_taylor_series(3, x)))
    error4.append(abs(np.sin(x) - sin_taylor_series(4, x)))
```

```python
plt.plot(x_val, error1, label='1 term')
plt.plot(x_val, error2, label='2 terms')
plt.plot(x_val, error3, label='3 terms')
plt.plot(x_val, error4, label='4 terms')
plt.xlabel('x')
plt.ylabel('abs error')
plt.title('truncation error for sin(x)')
plt.legend()
plt.grid(True)
plt.show()
```

The truncation error stays tiny when x is small, but as you move further out it shoots up really fast. For big x, adding more terms can actually make things worse at first because those higher powers get huge compared to the real sin value.
